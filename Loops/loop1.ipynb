```python
# den= denominator
# num= numerator

volume=[500,400,300]
prices=[100,200,300]

num=0
den=0
for i in range(len(volume)):
    num=num+(volume[i]*prices[i])
    den=den+volume[i]
print(num/den)
```

```python
volume=[500,400,300]
prices=[100,200,300]

num=0
den=0

index=0
while True:
    num=num+(volume[index]*prices[index])
    den=den+volume[index]

    index=index+1
    if index==len(volume):
        break
    print(num/den)
```
